# <font color="Green">***Code TF-IDF From Scratch***</font>

```python
import math
from collections import Counter

documents = [
    "cat sat on mat",
    "dog sat on mat",
    "cat likes fish"
]

# Step 1: Tokenize
tokenized_documents = [
    document.lower().split()
    for document in documents
]

# Step 2: Calculate TF
def calculate_tf(document):
    counts = Counter(document)
    total_words = len(document)

    tf = {}

    for word, count in counts.items():
        tf[word] = count / total_words

    return tf


# Step 3: Calculate IDF
def calculate_idf(documents):

    total_documents = len(documents)

    vocabulary = set(
        word
        for document in documents
        for word in document
    )

    idf = {}

    for word in vocabulary:

        document_count = sum(
            1 for document in documents
            if word in document
        )

        idf[word] = math.log(
            total_documents / document_count
        )

    return idf


# Calculate TF for each document
tf_values = [
    calculate_tf(document)
    for document in tokenized_documents
]

# Calculate IDF
idf_values = calculate_idf(tokenized_documents)

print("TF:")
print(tf_values)

print("\nIDF:")
print(idf_values)
```

```python
tfidf_values = []

for tf in tf_values:

    document_tfidf = {}

    for word, tf_score in tf.items():

        idf_score = idf_values[word]

        tfidf_score = tf_score * idf_score

        document_tfidf[word] = tfidf_score

    tfidf_values.append(document_tfidf)


print("TF-IDF:")

for document in tfidf_values:
    print(document)
```

```python
# Create Skip Gram
window_size=1
training_data=[]
for i in range(len(words)):
  center_word=words[i]
  start=max(0, i-window_size)
  end=min(len(words), i+window_size+1)
  for j in range(start, end):
    if j!=1:
      context_word=words[j]
      training_data.append((center_word, context_word))
print("training data.......... ")
```

```python
# model Parameters
embedding_size=5
learning_rate=0.1
epochs=100
# Weights
np.random.seed(42)
W=np.random.randn(V, embedding_size)*0.01
W_prime=np.random.randn(embedding_size, V)*0.01
# One Hot Function
def  one_hot(word):
  vector=np.zeros(V)
  vector[word_to_index[word]]=1
  return vector
# Softmax
def softmax(x):

    exp_x = np.exp(
        x - np.max(x)
    )

    return exp_x / np.sum(exp_x)

```

# <font color="blue">***GloVe From  Scratch***</font>

```python
import numpy as np
Sentence="I Love Machine Learning "
words=Sentence.lower().split()
words
```

```python
# Create Vocabulry
vocabulary=sorted(set(words))
vocabulary
```

```python
# Create Word Index Mapping
word_to_index={}
for index , word in enumerate(vocabulary):
  word_to_index[word]=index
print(word_to_index)
```

```python
# Create Empty Co-Occurance Matrix
V=len(vocabulary)
co_occurance_matrix=np.zeros((V, V))
print(co_occurance_matrix)

print("INIITIALLY EMPTY CO-OCCURANCE MATRIX WITHOUT NEIGHBORS")
```

```python
# set Context Window
window_size=1
```

```python
# 1. Reset the matrix to zeros every time you run this cell
co_occurance_matrix = np.zeros((V, V))

# 2. Build Co-occurrence Matrix
for index, word in enumerate(words):
    word_index = word_to_index[word]
    start = max(0, index - window_size)
    end = min(len(words), index + window_size + 1)

    for j in range(start, end):
        if index != j:
            context_word = words[j]
            context_index = word_to_index[context_word]
            co_occurance_matrix[word_index][context_index] += 1

# 3. Print the Correct Matrix
print("\nCo-occurrence Matrix\n")
print("      ", end="")
for word in vocabulary:
    print(f"{word:10}", end="")
print()

for i, word in enumerate(vocabulary):
    print(f"{word:6}", end="")
    for value in co_occurance_matrix[i]:
        print(f"{int(value):10}", end="")
    print()
```

#<font color="Orange">***Fasttext Developed by Facebook from Scratch***</font>

```python
import numpy as np
Sentence="I Love Machine Learning"
words=Sentence.lower().split()
words
```

```python
# Create Vocabulry
vocabulary=sorted(set(words))
vocabulary
```

```python
word_to_index={}
for index , word in enumerate(vocabulary):
  word_to_index[word]=index

```

```python
# Create N-Grams
# Let's Use 3-Grams
def get_ngrams(word , n=3):
  word= "<" + word + ">"
  ngrams=[]
  for i in range(len(word)-n+1):
    ngrams.append(word[i:i+n])
  return ngrams


```

```python
print(get_ngrams("learning"))
```

```python
# for EveryWord Get N-grams
word_ngrams={}
for word in vocabulary:
  word_ngrams[word]=get_ngrams(word)
print(word_ngrams)
```

```python
all_ngrams=set()
for ngrams in word_ngrams.values():
  for ng in ngrams:
    all_ngrams.add(ng)
  print(all_ngrams)
```

```python
sorted_ngrams=sorted(all_ngrams)
sorted_ngrams
```

```python
# Build INdex Dictoniary
ngram_to_index = {}
for i in range(len(sorted_ngrams)):
    ng = sorted_ngrams[i]
    ngram_to_index[ng] = i

print(ngram_to_index)
```

```python
embedding_size=5
np.random.seed(42)
ngram_embeddings=np.random.randn(len(ngram_to_index), embedding_size)
```

```python
def get_word_vector(word):

    ngrams = get_ngrams(word)

    vectors = []

    for ng in ngrams:
        if ng in ngram_to_index: # Check if ngram exists in our index
            index = ngram_to_index[ng]
            vectors.append(ngram_embeddings[index])

    if vectors: # Avoid division by zero if no ngrams are found
        return np.mean(vectors, axis=0)
    else:
        # Return a zero vector if no known ngrams are found for the word
        return np.zeros(embedding_size)
```
